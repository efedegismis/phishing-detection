```python
from xgboost import XGBClassifier
import time
import json
import sys
from pathlib import Path
project_root = Path('..').resolve()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
from src import data_loader
```

```python
X_train, X_val, X_test, y_train, y_val, y_test = data_loader.load_data()

xgboost_model = XGBClassifier(n_estimators=100, 
                              learning_rate=0.05, 
                              max_depth=5, 
                              objective="binary:logistic", 
                              early_stopping_rounds=15,
                              random_state=42)

start = time.time()
xgboost_model.fit(X_train, y_train, 
                  eval_set=[(X_val, y_val)],
                  verbose=False)
training_time = time.time() - start

print(f"XGBoost model successfully trained in {training_time:.2f}s")
```

```python
xgboost_model.save_model("../saved_models/xgboost_model_v1.json")
print("Model saved to saved_models/xgboost_model_v1.json")

times_path = '../saved_models/training_times.json'
try:
    with open(times_path, 'r') as f:
        times = json.load(f)
except FileNotFoundError:
    times = {}
times['XGBoost'] = training_time
with open(times_path, 'w') as f:
    json.dump(times, f)
print(f"Training time is saved to training_times.json")
```
