```python
import numpy as np
import pandas as pd
import sys
from pathlib import Path
project_root = Path('..').resolve()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))
from src import data_loader, feature_extractor
```

```python
X_train, X_val, X_test, y_train, y_val, y_test = data_loader.load_data()

X_inference, y_inference = feature_extractor.extract_features_from_dataset()

comparison = pd.DataFrame({
    "train_mean": np.mean(X_train, axis=0),
    "inference_mean": np.mean(X_inference, axis=0),
    "train_std": np.std(X_train, axis=0),
    "inference_std": np.std(X_inference, axis=0),
    "train_min": np.min(X_train, axis=0),
    "inference_min": np.min(X_inference, axis=0),
    "train_max": np.max(X_train, axis=0),
    "inference_max": np.max(X_inference, axis=0),
})

print(comparison.round(4).to_string())
```

```python
feature_names = [
    "URLLength", "DomainLength", "IsDomainIP",
    "CharContinuationRate", "TLDLength", "NoOfSubDomain",
    "HasObfuscation", "NoOfObfuscatedChar", "ObfuscationRatio",
    "NoOfLettersInURL", "LetterRatioInURL", "NoOfDegitsInURL",
    "DegitRatioInURL", "NoOfEqualsInURL", "NoOfQMarkInURL",
    "NoOfAmpersandInURL", "NoOfOtherSpecialCharsInURL",
    "SpacialCharRatioInURL", "IsHTTPS"
]

train_df = pd.DataFrame(X_train, columns=feature_names)
train_df["label"] = np.asarray(y_train)

inference_df = pd.DataFrame(X_inference, columns=feature_names)
inference_df["label"] = np.asarray(y_inference)

print("Training means by class:")
print(train_df.groupby("label")[feature_names].mean().T.round(3))

print("\nInference means by class:")
print(inference_df.groupby("label")[feature_names].mean().T.round(3))
```
